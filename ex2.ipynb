a) Distribuția numărului de pași N
Numărul de pași
 până la încheierea jocului reprezintă numărul de aruncări ale monedei până la prima apariție a stemei. Deoarece aruncările sunt independente, iar probabilitatea de a obține stemă este constantă la fiecare pas, variabila aleatoare N
 urmează o distribuție geometrică cu parametrul p
 (probabilitatea succesului).

b) Simularea jocului în Python
Funcția de mai jos simulează o rundă de joc. La fiecare pas, se aruncă moneda. Dacă pică ban, al doilea jucător trebuie să primească 0.5$, ceea ce echivalează matematic cu o scădere a sumei S pe care el trebuie să o dea primului jucător (S scade cu 0.5). Dacă pică stemă, zarul determină valoarea z, iar la S se adaugă (z - 3), moment în care jocul se termină. Se returnează numărul de pași și suma finală S.

```python
import numpy as np
import matplotlib.pyplot as plt

def simuleaza_joc(p=0.5):
    n = 0
    s = 0.0

    while True:
        n += 1
        picat_stema = np.random.rand() < p

        if picat_stema:
            z = np.random.randint(1, 7)
            s += (z - 3)
            break
        else:
            s -= 0.5

    return n, s

n_test, s_test = simuleaza_joc()
print(f"Exemplu un joc (p=0.5): N = {n_test}, S = {s_test} $")
```

c) Simularea unui număr mare de jocuri și histogramele pentru p=0.5
Se realizează o iterație de 10.000 de jocuri folosind o monedă echilibrată. Vom calcula media cu aproximație a sumei
 (teoretic ar trebui să fie 0) și vom trasa distribuția utilizând o histogramă.

```python

numar_simulari = 10000
rezultate_s = []

for _ in range(numar_simulari):
    _, s = simuleaza_joc(p=0.5)
    rezultate_s.append(s)

media_s = np.mean(rezultate_s)
print(f"Media lui S pentru p=0.5 este: {media_s:.4f} $")

plt.figure(figsize=(8, 5))
plt.hist(rezultate_s, bins=range(int(min(rezultate_s)), int(max(rezultate_s)) + 2), align='left', rwidth=0.8, color='skyblue', edgecolor='black')
plt.title("Distribuția sumei S (p = 0.5)")
plt.xlabel("Suma S ($)")
plt.ylabel("Frecvență")
plt.grid(axis='y', alpha=0.75)
plt.show()
```

d) Efectul unei monede măsluite (p=0.3 și p=0.7)
Dacă moneda este măsluită, șansa obținerii stemei guvernează direct durata jocului. La p=0.3, este mult mai greu să apară stema, ceea ce crește numărul mediu de pași. Din această cauză, vor avea loc mai multe penalizări de 0.5$, ducând așteptarea totală a lui S pe minus. O valoare de p=0.7 scurtează jocul, rezultând mai puține penalizări și o valoare medie pozitivă.

```python
rezultate_s_03 = []
rezultate_s_07 = []

for _ in range(numar_simulari):
    _, s = simuleaza_joc(p=0.3)
    rezultate_s_03.append(s)

for _ in range(numar_simulari):
    _, s = simuleaza_joc(p=0.7)
    rezultate_s_07.append(s)

print(f"Media lui S pentru p=0.3 este: {np.mean(rezultate_s_03):.4f} $")
print(f"Media lui S pentru p=0.7 este: {np.mean(rezultate_s_07):.4f} $")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(rezultate_s_03, bins=25, color='salmon', edgecolor='black')
axes[0].set_title("Distribuția sumei S (p = 0.3)")
axes[0].set_xlabel("Suma S ($)")
axes[0].set_ylabel("Frecvență")
axes[0].grid(axis='y', alpha=0.75)

axes[1].hist(rezultate_s_07, bins=25, color='lightgreen', edgecolor='black')
axes[1].set_title("Distribuția sumei S (p = 0.7)")
axes[1].set_xlabel("Suma S ($)")
axes[1].set_ylabel("Frecvență")
axes[1].grid(axis='y', alpha=0.75)

plt.tight_layout()
plt.show()
```
